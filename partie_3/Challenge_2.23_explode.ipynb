- Which actor is mentioned the most often?


```
import pandas as pd

link = 'https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/imdb_1000.csv'
data = pd.read_csv(link)
```

```
import ast

data["actors_list"] = data["actors_list"].apply(ast.literal_eval)

data["actors_list"].explode().value_counts().head(1)
```
