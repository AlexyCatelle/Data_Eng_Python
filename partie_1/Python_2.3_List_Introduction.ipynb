2. What can the `.split()` method do?  
What is the type of `christmas` before using `.split()`?  
What is the type of `christmas` after using `.split()`?

Answer:
The `.split()` method splits a string into a list of substrings, using spaces as the default separator.
Before using `.split()`, `christmas` is a `str`.
The result of `.split()` is a `list`. However, `christmas` remains a `str` unless the result is assigned back to the variable.

```python
christmas = "Noël c'est bien, mais attention à ne pas manger trop de chocolat. On le sait, mais on le fait à chaque fois. C'est la vie"


christmas.split("à")
```

3. Why does the `.split()` method split the `string` into 3 elements?

Answer: The `.split("à")` method splits the string at each occurrence of the character `"à"`. Since `"à"` appears twice in the string, the result contains three elements.
