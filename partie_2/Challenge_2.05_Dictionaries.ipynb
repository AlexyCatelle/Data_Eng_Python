# Mission 1

Below is a dictionary containing the number of French campuses that offer each kind of Wild Code School course.

wildcodeschool = {
   "data_analyst": 4,
   "data_scientist": 2,
   "dev_web": "13"
}

Find a way to only display the keys that have int values.

```python
wildcodeschool =	{
  "data_analyst": 4,
  "data_scientist": 2,
  "dev_web": "13"
}


for key, value in wildcodeschool.items():
    if isinstance(value, int):
        print(key)
```

# Mission 2

For the same dictionary, find a way to display all the values that have a key that contains the letter "e".

```python
wildcodeschool =	{
  "data_analyst": 4,
  "data_scientist": 2,
  "dev_web": "13"
}

for key, value in wildcodeschool.items():
    if "e" in key:
        print(key," : ", value)
```

# Mission 3

```python

# Change the type's value of the key "dev_web" to integer
wildcodeschool["dev_web"] = int(wildcodeschool['dev_web'])

#Check the type
print(type(wildcodeschool["dev_web"]))

```

```python
# add a new key-value pair which will be "total_campus" : 19
wildcodeschool["total_campus"] = 19


print(wildcodeschool)
```

# Mission 6

Create a script that returns the same elements as the **values()** method, as a list.

```python
dico = {'a': 1, 'b': 2, 'c': 3}

valueList = []

for value in dico :
    valueList.append(dico[value])

print(valueList)
```
