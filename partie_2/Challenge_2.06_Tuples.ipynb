# Mission 4

Create 2 tuples of the same length that contain only integers.

Then create a script that compares the sum of their elements and displays the tuple that has the highest total.

```python
# Numéros des pilotes déjà champions du monde
champions = (1, 3, 14, 44)

# Numéros de quatre autres pilotes
autres = (5, 10, 16, 81)

# Calculer les sommes
total_champions = sum(champions)
total_autres = sum(autres)

# Comparer les sommes
if total_champions > total_autres:
    print("Les champions ont le total le plus élevé.")
elif total_autres > total_champions:
    print("Les autres pilotes ont le total le plus élevé.")
else:
    print("Les deux tuples ont le même total.")
```

# Mission 5

Here's a tuple:

`my_tuple = ("data analyst", "data scientist", "data engineer", "data architect")`

Ask the user for a string, then for an integer position. Create an altered copy of `my_tuple` where the element at the given position is now the given string.

```python
my_tuple = ("data analyst", "data scientist", "data engineer", "data architect")

# Ask the user for a string
new_value = input("Nouveau string : ")

# then for an integer position
new_value_pos = int(input("position : "))

# Create an altered copy of `my_tuple` where the element at the given position is now the given string

alt_my_tuple = my_tuple[:new_value_pos] + (new_value ,) + my_tuple[ new_value_pos:]

print(alt_my_tuple)
```
