# Mission 2

Create a function that returns the sum of two numbers.  
The user of the function must be able to sum two numbers of his choice, when the function is executed.  
You must not use `input`.  


```python
def sum_ab () :
    a = int(input("première valeur : "))
    b = int(input("deuxieme valeur : "))
    
    sum = a + b
    
    return sum

sum_ab()
```

What is a parameter in a function?
Un paramètre (parameter) est une variable définie entre les parenthèses lors de la création d'une fonction. Il permet de transmettre une information à la fonction pour qu'elle puisse l'utiliser.

# Mission 4

Create a function that returns 2 objects when executed:
* An object that will be the product of two numbers chosen by the user of the function.
* A word chosen by the user of the function.

Your result must be two objects that are in brackets.  
This is a tuple.

Question:  
Why is the returned object a tuple that contains two objects?  

Answer:
Because in Python, when you return two objects separated by a comma, Python automatically packs them into a tuple. Parentheses are optional when creating a tuple.

```python
def tuple_maker() :
    # An object that will be the product of two numbers chosen by the user of the function.
    a = float(input("nombre n°1 : "))
    b = float(input("nombre n°2 : "))

    p = a*b

    # A word chosen by the user of the function.
    w = input("mot : ")

    return p,w

print(tuple_maker())
print (type(tuple_maker()))
```

# Mission 5

Create a function that returns the sum of two numbers when executed.  
Also, your function must return a sentence that simply indicates if the sum of the two numbers chosen by the user is:
* less than 10
* between 10 and 50
* greater than 50  
Again, you must use `return`.

```python
def sum_if():
    a = float(input("nombre n°1 : "))
    b = float(input("nombre n°2 : "))

    s = a + b

    r = ""
    # less than 10
    if s < 10 :
        r = "less than 10"

    # between 10 and 50
    elif s >= 10 and s <= 50 :
        r = "between 10 and 50"
    
    # greater than 50
    else :
        r = "greater than 50"

    return f'The sum of the two numbers is {r}.'


print(sum_if())
```
